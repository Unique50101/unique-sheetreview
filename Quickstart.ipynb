```python
from pathlib import Path
from tempfile import mkdtemp
from unique_sheetreview import scan
from openpyxl import load_workbook

folder = Path(mkdtemp())
source = folder / "orders.csv"
source.write_text("OrderID,OrderDate,Customer\n101,04/05/2026, Asha\n101,2026-05-04,Amir\n,2026-05-05,Sam\n", encoding="utf-8")
report = scan(source, required=["OrderID"], unique=["OrderID"])
print(report)
review = report.save_review_copy(folder / "orders_review.xlsx")
assert len(report.findings) == 5 and load_workbook(review).active["A2"].comment
print("Review workbook:", review)
```
